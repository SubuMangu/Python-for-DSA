```python
mat=[1,2,3]+[4,5]
print(mat)
```

- We intialise empty list by `[]`

```python
empty_list=[]
if empty_list==[]:
    print("Empty list")
```

- Python lists list does not support multidimentional slicing shown below, it is supported by numpy arrays and pytorch tensors. 

```python
# Let's make three lists
lst_1=[1,2,3]
lst_2=[4,5,6]
lst_3=[7,8,9]

# Make a list of lists to form a matrix
matrix = [lst_1,lst_2,lst_3]
print(matrix)
indexed_list=matrix[:,0]
print(indexed_list)
```

- We use this instead or we can use list comprehensions for most customized way

```python
indexed_list=matrix[0][0]
print(indexed_list)
```

**Stride**
- We can use strides in a list as shown below

```python
nums=[1,2,3,4,5,6,7,8,9]
stride=2
selected_nums=nums[1::stride]
print(selected_nums)
```

- When the stride is negetive, it moves in reverse direction starting with last index.

```python
stride=-3
selected_nums_inv=nums[::stride]
print(selected_nums_inv)
```

## 2.String
- Indexing,stride and concatination is similar to list
- We will discuss some special methods used on strings:
    - substring
    - split
    - lower and upper
    - f strings or customized strings

**Substring**
- It is imp in DSA pov.
- We do by simply indexing. 

```python
word="Hello"
sub_word=word[:3]
print(sub_word)
```

**Split**
- By default it splits blank spaces.
- We can add strings as separator

```python
sentence="Hello! How are you?"
splited_words_1=sentence.split()
print("Default splitting: ",splited_words_1)
splited_words_2=sentence.split("are")
print("Customized split: ",splited_words_2)
```

**lower and upper**

```python
sentence2="Hello! Are you fine?"
print(sentence2.upper())
print(sentence2.lower())
```

**f strings**
- Use to insert values in between sentences

```python
amount=5
sentence3=f"I have {amount} bananas"
print(sentence3)
```

- We initialise empty string by `""` or `''`.

```python
empty_string=""
if empty_string == "":
    print("String is empty")
```

## 3. Dictionary
- In DSA it is used for hashmapping

```python
my_dict = {'key1':'value1','key2':'value2'}
# Call values by their key
my_dict['key2']
```

- We add new key value pair by directly assigning as shown below:

```python
my_dict['key3']='value3'
print(my_dict)
```

- `keys()`: returns keys in a dictionary
- `values()`: returns values in a dictionary
- `items()`: returns key value pair tuples

```python
print(my_dict.keys())
print(my_dict.values())
print(my_dict.items())
```

- We can iterate in dictionary by traversing `keys()`,`values()` or `items()`.

- We use `pop` to delete a key value, returns the value taking key as the input.

```python
my_dict.pop("key3", None)  # Returns None if "key3" doesn't exist
print(my_dict)
```

- We initialise empty dictionary by `{}`

```python
empty_dict={}
if empty_dict=={}:
    print("Dictionary is empty")
print(type(empty_dict))
```

## 4.Sets
- Represents a collection of unordered unique items

```python
items={1,2,3}
type(items)
```

- We add and remove item using `add` and `remove` functions. These functions return `None`

```python
items.add(10)
print(items)
items.remove(10)
print(items)
```

- We can traverse by using simple `for` loop

```python
for item in items:
    print(item)
```

- We represent empty set by `set()`

```python
empty_set=set()
if empty_set==set():
    print("Set is empty")
```

- We can convert a list with multiple number to a set with unique numbers by passing the list to `set()`

```python
# Create a list with repeats
list1 = [1,1,2,2,3,4,5,6,1,1]
# Cast as set to get unique values
set(list1)
```
