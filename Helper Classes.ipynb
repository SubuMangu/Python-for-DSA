# Helper Class
- It is useful in DSA for solving problems on Tree,Queues,Priority Queues, etc, where there is a need to create Abstract Datatypes.
- There are 2 types of problems involving helper classes:
    1. Where helper classes are explicity mentioned(Type 1)
    2. Where we have to create helper class(Type 2)

```python
from typing import Optional


class ListNode:
    def __init__(self,val=0,next=None):
        self.val=val
        self.next=next
    
ptr0=ListNode(1)
result=ptr0

ptr0.next=ListNode(2)
ptr0=ptr0.next

ptr0.next=ListNode(3)
ptr0=ptr0.next

def print_listnode(list:Optional[ListNode]):
    while list is not None:
        print(f'{list.val} -->',end=" ")
        list=list.next
    print("NULL")

print_listnode(result)

    
```
