# Scope 
- Python follows LEGB rule to find to the value of a variable 
1. **L(Local):** Variables inside function under consideration
2. **E(Enclosing):** Variables in function enclosing the function under consideration
3. **G(Global):** Variables declared outside of all functions
4. **B(Build in):** Not required for DSA

```python
# Globa Scope
x="global"
class Dummy():
    def fun():
        # Enclosing Scope
        x="enclosing"
        def inner():
            # Local Scope
            x="local"
            print(x)
        inner()
        print(x)

Dummy.fun()
print(x)
```

**`nonlocal`  keyword**
- It is used to
    1. Modify immutable enclosing variables like `int`,`float`,`string` and `boolean`
    2. Used to reassign enclosing mutable variables like `list`,`dict` and `set`

```python
class Dummy2():
    def fun():
        count=0
        item={2}
        def inner():
            nonlocal count
            nonlocal item
            count+=1
            item={0}
        inner()
        return count,item
count,item=Dummy2.fun()
print(count)
print(item)
```

**`global` keyword**
- It is used to
    1. Modify immutable global variables like `int`,`float`,`string` and `boolean`
    2. Used to reassign global mutable variables like `list`,`dict` and `set`

```python
count=0
item={2}
class Dummy3():
    def fun():      
        def inner():
            global count
            global item
            count+=1
            item={0}
        inner()
        return count,item
count,item=Dummy3.fun()
print(count)
print(item)
```

- **NOTE:** Parameters passed inside a function act as local variable inside that function
- In the code given below `count` act as local variable for `fun` but enclosing variable for `inner`

```python
from typing import List


class Dummy4():
    def fun(self,count:List[int]):
        def inner():
            count[0]+=1
        inner()
        return count
ob=Dummy4()
val=ob.fun([0,1])
print(val)
```
