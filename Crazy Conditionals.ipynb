# Crazy conditionals
- Datatypes other than boolean can be used as condition

### Objects as conditions
- If the object used in condition in `None`, it wouldn't execute the statement below the condition, otherwise it would.

```python
class World():
    def __init__(self,val:int=0):
        self.val=val
null_object=None
object=World(1)
if null_object:
    print("Null object")
if object:
    print("Object")
```

### Numbers as conditions
- If the value used as condition is zero, statements are not executed.

```python
## Float as condition
val1=0.0
val2=0.1
if val1:
    print("Hello val1")
if val2:
    print("Hello val2")
# Integer as condition
int1=0
int2=17
if int1:
    print("Hello int1")
if int2:
    print("Hello int2")
```

### Strings as conditions

```python
empty_string=""
word="Hello"
if empty_string:
    print("Empty string printed")
if word:
    print("Non empty string printed")
```

### Dictionary as condition

```python
empty_map={}
map={1:"Hello"}
if empty_map:
    print("Empty map printed")
if map:
    print("Map is printed")
```

### Set as condition

```python
empty_set=set()
items={1,2}
if empty_set:
    print("Empty set printed")
if items:
    print("Items are printed")
```

### Tuple as condition

```python
empty_tuple=()
tuple=(1,2)
if empty_tuple:
    print("Empty tuple is printed")
if tuple:
    print("Tuple is printed")
```
