# Operators
- The important operators used in python are given below.

**Integer Division**

```python
7//4
```

**Exponent**

```python
# Powers
2**3
```

**Priority order of operators**
- Python follows **PEMDAS** rules
    - **P**arentheses `()`

    - **E**xponents `**`

    - **MD** Multiplication `*`, Division `/`, Floor Division `//`, Modulus `%`

    - **AS** Addition `+`, Subtraction `-`

```python
result = 10 + 2 * 3 ** 2
print(result)
```

Step-by-step:
1. `3 ** 2` = `9`
2. `2 * 9` = `18`
3. `10 + 18` = `28`

✔️ Output: 28

```python
result = (10 + 2) * 3 ** 2
print(result)
```

**Multiple Assignment**
- We can do multiple assignments in one like

```python
a=b=10
print(a)
print(b)
```

**Self Operation**

```python
a=1
a=a+1
print(a)
a+=1
print(a)
```

- There is no unary operator used for increament or decreament in python like `a++`, which is commonly used in other programming languages

**Mulitple Assignment in Tuples**

```python
x,y=(1,2)
print(x)
print(y)
```
