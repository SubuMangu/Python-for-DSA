# Other Data Structure
- Here we will discuss about the Data Structure that is not build in in python.

## 1.Stack 

```python
stack=[]
# pushing values
stack.append(1)
stack.append(2)
stack.append(3)
print("Stack before deletion: ",stack)
# poping values
item=stack.pop()
print("Stack after deletion: ",stack)
print("Removed item: ",item)
```

## 2.Queue

```python
queue=[]
# Enque
queue.append(1)
queue.append(2)
queue.append(3)
print("Queue before deletion: ",queue)
# Deque
item=queue.pop(0)
print("Queue after deltion: ",queue)
print("Removed item: ",item)
```

## 3.Deque

```python
from collections import deque

# Initialize a deque (optionally with an iterable)
dq = deque([2, 3, 4])

# Add to the right side
dq.append(5)          # deque([2, 3, 4, 5])

# Add to the left side
dq.appendleft(1)      # deque([1, 2, 3, 4, 5])
# Remove and return from the right side
right_item = dq.pop()      # returns 5 -> deque([1, 2, 3, 4])

# Remove and return from the left side
left_item = dq.popleft()   # returns 1 -> deque([2, 3, 4])
```
