# Call by Reference
- Call by reference means, the method(to which the variable is passed) or variable(to which the value is assigned) gets the address of the value.
- By default in python, in all mutable datatypes are called by their reference.

```python
# Call by reference in case of functions
def insert(arr:list,val:int):
    arr.append(val)
arr1=[]
insert(arr1,1)
print(arr1) # Prints the inserted value
```

```python
# Call by reference in case of variables
arr1=[]
print("Arr1: ",arr1)
arr2=arr1
arr1.append(1)# Also got inserted in arr2
print("Arr2: ",arr2)
```

- Since self made datatypes (as shown below as `ListNode`) are mutable variable, they follow call by reference by default. 
- Hence, we can see that  `ListNode(2)` also got added to `result` since it stores the reference of `ptr0` 

```python
from typing import Optional


class ListNode:
    def __init__(self,val=0,next=None):
        self.val=val
        self.next=next
    
ptr0=ListNode(1)
result=ptr0

ptr0.next=ListNode(2)## This is also saved in result
ptr0=ptr0.next



def print_listnode(list:Optional[ListNode]):
    while list is not None:
        print(f'{list.val} -->',end=" ")
        list=list.next
    print("NULL")

print_listnode(result)
```

- Immutable variables have no call by reference, but we change the value by reassigning.

```python
def increament(val:int):
    return val+1
count=0 # Immutable variable
count=increament(count) # Changing value by reassigning
print(count)
```
