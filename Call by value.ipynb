# Call by Value
- Call by Value means, the method(where the variable is passed) or the variable(to which the value is assigned) gets the copy of the value of the variable passed.
- All the immutable are called by value.

```python
# Call by value in case of functions
def increament(val:int):
    val+=1
count=0
increament(count)
print(count) # No change in value
```

```python
# Call by value in case of variable
num1=1
num2=num1
num2+=1
print(num1)# No change in value of num1
```

- In DSA problems like backtracking, where we need to recursively call sometimes the mutable variables. As we reach the first endpoint, we can't backtrack just because we can't restore the value, since they are modified as their references are passed by default.
- So, to restore the value of mutable variables in reccursive problems, we can pass a copy of the mutable variable using `.copy()` as shown below.

```python
def delete(arr:list):
    if arr==[]:
        return
    arr.pop()
arr=[1,2]
delete(arr.copy()) # Passing a copy of arr to restore values
print(arr)
```
